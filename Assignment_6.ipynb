```python
from sklearn.datasets import fetch_california_housing
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

housing = fetch_california_housing()
df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['MedHouseVal'] = housing.target
print(df.head())
print(df.isnull().sum())
X = df.drop('MedHouseVal', axis=1)
y = df['MedHouseVal']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
print(X_scaled.head())

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
lr = LinearRegression()
dt = DecisionTreeRegressor(random_state=42)
rf = RandomForestRegressor(random_state=42)
gb = GradientBoostingRegressor(random_state=42)
svr = SVR()

lr.fit(X_train, y_train)
dt.fit(X_train, y_train)
rf.fit(X_train, y_train)
gb.fit(X_train, y_train)
svr.fit(X_train, y_train)

models = {"Linear Regression": lr, "Decision Tree": dt, "Random Forest": rf, "Gradient Boosting": gb, "SVR": svr}
results = []
for name, model in models.items():

    y_pred = model.predict(X_test)

    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    results.append([name, mse, mae, r2])

results_df = pd.DataFrame(results, columns=["Model", "MSE", "MAE", "R² Score"])

print(results_df.sort_values(by="R² Score", ascending=False))
```

Explanation of Preprocessing Steps:
1.Conversion to DataFrame
.The dataset was converted into a Pandas DataFrame to simplify data exploration, manipulation, and analysis.
2.Missing Value Check
.Missing values can negatively affect machine learning models.
.The California Housing dataset contains no missing values, so no imputation was needed.
3.Feature Scaling (Standardization)
.Features in the dataset have different ranges (e.g., population values are much larger than median income values).
.Standardization transforms features to have a mean of 0 and a standard deviation of 1.
.This helps regression algorithms such as Support Vector Regression (SVR), Gradient Boosting, and Linear Regression perform more efficiently and prevents features with larger scales from dominating the model.

1. Linear Regression
.Simple and easy to interpret.
.Provides a baseline model for comparison.
.Works well when relationships between features and target are approximately linear.
.Fast to train even on large datasets.

2. Decision Tree Regressor
.Captures non-linear relationships.
.Easy to understand and visualize.
.Handles complex interactions between housing features.
.Does not require feature scaling.

3. Random Forest Regressor
.Reduces overfitting compared to a single decision tree.
.Produces more accurate and stable predictions.
.Handles large datasets efficiently.
.Works well with housing datasets containing multiple features.

4. Gradient Boosting Regressor
.Excellent predictive performance.
.Captures complex non-linear patterns.
.Handles feature interactions effectively.
.Frequently used in real-world regression problems such as house price prediction.

5. Support Vector Regressor (SVR)
.Effective for high-dimensional datasets.
.Can model complex non-linear relationships using kernel functions.
.Less sensitive to outliers than some regression methods.
.Works well when features are standardized.

Best-Performing Algorithm: Random Forest Regressor

Justification:

Achieved the lowest MSE and lowest MAE, indicating the smallest prediction errors.
Produced the highest R² score, meaning it explained the largest proportion of variance in housing prices.
Combines multiple decision trees, reducing overfitting and improving generalization.
Effectively captures complex relationships within the California Housing dataset.
Worst-Performing Algorithm: Linear Regression

Reasoning:

Had the highest MSE and MAE among the models.
Produced the lowest R² score, indicating weaker predictive performance.
Assumes a linear relationship between features and house prices, whereas the California Housing dataset contains several non-linear relationships.
Unable to capture complex feature interactions as effectively as tree-based ensemble methods.
